# Section-1: Creating the Database and SQL Schema


```python
"""
schema.py
=========
Defines and applies the hospital analytics database schema
entirely in Python using SQLite.

Tables:   patients → visits → billing
Output:   hospital.db
"""
DB_PATH = Path("hospital.db")

# ── DDL Statements ────────────────────────────────────────────────────────────

CREATE_PATIENTS = """
CREATE TABLE IF NOT EXISTS patients (
    patient_id          INTEGER     PRIMARY KEY,
    age                 INTEGER     NOT NULL CHECK (age >= 0 AND age <= 130),
    gender              VARCHAR(20) NOT NULL,
    city                VARCHAR(100),
    insurance_provider  VARCHAR(100),
    chronic_flag        INTEGER     NOT NULL DEFAULT 0 CHECK (chronic_flag IN (0, 1)),
    registration_date   DATE        NOT NULL
);
"""

CREATE_VISITS = """
CREATE TABLE IF NOT EXISTS visits (
    visit_id                INTEGER         PRIMARY KEY,
    patient_id              INTEGER         NOT NULL,
    visit_date              DATE            NOT NULL,
    department              VARCHAR(100)    NOT NULL,
    visit_type              VARCHAR(50)     NOT NULL,
    length_of_stay_hours    DECIMAL(8, 2),
    risk_score              VARCHAR(20),
    doctor_id               INTEGER,

    FOREIGN KEY (patient_id)
        REFERENCES patients (patient_id)
        ON DELETE CASCADE
        ON UPDATE CASCADE
);
"""

CREATE_BILLING = """
CREATE TABLE IF NOT EXISTS billing (
    bill_id             INTEGER         PRIMARY KEY,
    visit_id            INTEGER         NOT NULL,
    billed_amount       DECIMAL(12, 2)  NOT NULL CHECK (billed_amount >= 0),
    approved_amount     DECIMAL(12, 2)  CHECK (approved_amount >= 0),
    claim_status        VARCHAR(50),
    payment_days        INTEGER         CHECK (payment_days >= 0),
    billing_date        DATE            NOT NULL,

    FOREIGN KEY (visit_id)
        REFERENCES visits (visit_id)
        ON DELETE CASCADE
        ON UPDATE CASCADE
);
"""

# ── Index Statements ──────────────────────────────────────────────────────────

INDEXES = [
    # patients
    "CREATE INDEX IF NOT EXISTS idx_patients_insurance   ON patients (insurance_provider);",
    "CREATE INDEX IF NOT EXISTS idx_patients_city        ON patients (city);",
    "CREATE INDEX IF NOT EXISTS idx_patients_chronic     ON patients (chronic_flag);",
    # visits
    "CREATE INDEX IF NOT EXISTS idx_visits_visit_date    ON visits (visit_date);",
    "CREATE INDEX IF NOT EXISTS idx_visits_department    ON visits (department);",
    "CREATE INDEX IF NOT EXISTS idx_visits_patient_id    ON visits (patient_id);",
    "CREATE INDEX IF NOT EXISTS idx_visits_visit_type    ON visits (visit_type);",
    "CREATE INDEX IF NOT EXISTS idx_visits_risk_score    ON visits (risk_score);",
    # billing
    "CREATE INDEX IF NOT EXISTS idx_billing_claim_status ON billing (claim_status);",
    "CREATE INDEX IF NOT EXISTS idx_billing_date         ON billing (billing_date);",
    "CREATE INDEX IF NOT EXISTS idx_billing_visit_id     ON billing (visit_id);",
]

# ── Ordered list for easy iteration ──────────────────────────────────────────

TABLES = [
    ("patients", CREATE_PATIENTS),
    ("visits",   CREATE_VISITS),
    ("billing",  CREATE_BILLING),
]


def apply_schema(conn: sqlite3.Connection) -> None:
    """Create all tables and indexes on the given connection."""
    conn.execute("PRAGMA foreign_keys = ON;")

    for table_name, ddl in TABLES:
        conn.execute(ddl)
        print(f"  ✔ Table '{table_name}' created.")

    for idx_sql in INDEXES:
        conn.execute(idx_sql)

    conn.commit()
    print(f"  ✔ {len(INDEXES)} indexes created.")


def main() -> None:
    print(f"\n{'='*50}")
    print("  Applying Hospital Analytics Schema")
    print(f"{'='*50}\n")

    conn = sqlite3.connect(DB_PATH)
    apply_schema(conn)
    conn.close()

    print(f"\n  Database ready: {DB_PATH.resolve()}\n")


main()
```

# Section 2: Loading Dataset

# synthetic data generation

```python
# ══════════════════════════════════════════════════════════════════════════════
# SYNTHETIC DATA GENERATOR — Hospital Operations & Revenue Risk Intelligence
#
# ⚠️  THIS DATA IS SYNTHETIC AND ENGINEERED, NOT A CLEANED VERSION OF REAL DATA.
#
# Purpose: produce a portfolio-ready dataset where risk_score and claim_status
# have real, learnable, DOCUMENTED relationships with other fields, so a model
# trained on it can legitimately reach ~90%+ accuracy. Every injected
# relationship is listed in SYNTHETIC_DATA_README.md — nothing here is hidden
# or covered up. This is a common, transparent ML-portfolio practice for
# demonstrating a modeling/deployment pipeline separately from real-world
# messy-data work.
#
# All dates are generated so they are ALWAYS logically consistent:
#   registration_date <= visit_date <= billing_date, and billing_date is
#   always on/after an estimated discharge date.
# ══════════════════════════════════════════════════════════════════════════════

import numpy as np
import pandas as pd

rng = np.random.default_rng(42)

N_PATIENTS = 5000
N_VISITS = 25000

CITIES = ["Hyderabad", "Pune", "Chennai", "Bangalore", "Mumbai", "Delhi"]
INSURANCE_PROVIDERS = ["SecureLife", "HealthPlus", "CareOne", "MediCareX"]
DEPARTMENTS = ["Cardiology", "Orthopedics", "ICU", "General", "ER", "Neurology"]
VISIT_TYPES = ["ER", "OPD", "ICU"]

# ─────────────────────────────────────────────────────────────────────────────
# 1. PATIENTS
# Injected signal: chronic_flag probability rises with age (realistic and
# also used later to drive risk_score).
# ─────────────────────────────────────────────────────────────────────────────
patient_id = np.arange(1, N_PATIENTS + 1)
age = rng.integers(1, 91, size=N_PATIENTS)
gender = rng.choice(["M", "F"], size=N_PATIENTS)
city = rng.choice(CITIES, size=N_PATIENTS)
insurance_provider = rng.choice(INSURANCE_PROVIDERS, size=N_PATIENTS)

chronic_prob = np.clip(0.15 + (age / 90) * 0.55, 0.05, 0.85)
chronic_flag = rng.binomial(1, chronic_prob)

reg_start = pd.Timestamp("2025-01-20")
registration_date = reg_start + pd.to_timedelta(rng.integers(0, 200, size=N_PATIENTS), unit="D")

patients = pd.DataFrame({
    "patient_id": patient_id,
    "age": age,
    "gender": gender,
    "city": city,
    "insurance_provider": insurance_provider,
    "chronic_flag": chronic_flag,
    "registration_date": registration_date,
})

# ─────────────────────────────────────────────────────────────────────────────
# 2. VISITS
# Injected signal for risk_score (Low / Medium / High):
#   score = 1.6*chronic_flag + 1.2*(age>60) + 1.4*(visit_type=='ICU')
#           + 1.0*(department in ['ICU','Cardiology']) + noise
#   -> thresholded into 3 classes, THEN ~8% of labels are randomly flipped
#      to one neighbouring class to keep it realistic (no dataset is
#      perfectly separable) and to avoid a suspiciously perfect model.
#
# length_of_stay_hours is then generated CONDITIONAL on risk_score
# (High -> longer stays), which becomes another legitimate predictive feature.
# ─────────────────────────────────────────────────────────────────────────────
visit_patient_id = rng.choice(patient_id, size=N_VISITS)
visit_df = patients.set_index("patient_id").loc[visit_patient_id].reset_index()

visit_id = np.arange(1, N_VISITS + 1)
department = rng.choice(DEPARTMENTS, size=N_VISITS)
visit_type = rng.choice(VISIT_TYPES, size=N_VISITS, p=[0.35, 0.45, 0.20])
doctor_id = rng.integers(101, 201, size=N_VISITS)

# visit_date always on/after that patient's registration_date
days_after_reg = rng.integers(0, 250, size=N_VISITS)
visit_date = visit_df["registration_date"].values + days_after_reg.astype("timedelta64[D]")

# A nurse-triage-style vitals flag captured at admission — an independent
# clinical signal, causally related to the same underlying acuity drivers,
# but NOT derived from risk_score. Simulates a realistic additional feature
# a hospital EHR would actually capture.
chronic_v = visit_df["chronic_flag"].values
age_v = visit_df["age"].values
vitals_prob = np.clip(0.10 + 0.35 * chronic_v + 0.25 * (age_v > 60) + 0.30 * (visit_type == "ICU"), 0.05, 0.9)
vitals_abnormal_flag = rng.binomial(1, vitals_prob)

# risk_score driven by clinical acuity factors (chronic condition, age,
# visit type, department, vitals flag) + noise, then thresholded into 3
# classes (same style as a real triage scoring rubric).
score = (
    1.9 * chronic_v
    + 1.3 * (age_v > 60).astype(float)
    + 1.7 * (visit_type == "ICU").astype(float)
    + 1.2 * np.isin(department, ["ICU", "Cardiology"]).astype(float)
    + 1.9 * vitals_abnormal_flag
    + rng.normal(0, 0.24, size=N_VISITS)
)
q1, q2 = np.quantile(score, [0.435, 0.435 + 0.314])
risk_score = np.where(score <= q1, "Low", np.where(score <= q2, "Medium", "High"))

# ~3% random label noise (keeps it realistic, avoids a suspiciously perfect model)
flip_mask = rng.random(N_VISITS) < 0.03
noise_alt = rng.choice(["Low", "Medium", "High"], size=N_VISITS)
risk_score = np.where(flip_mask, noise_alt, risk_score)

# length_of_stay_hours generated from the SAME underlying causal drivers as
# risk_score (not from the risk_score label itself) + its own noise term —
# this is what makes it a legitimate correlated predictor rather than a
# disguised copy of the target.
los_base = (
    10
    + 9 * chronic_v
    + 6 * (age_v > 60).astype(float)
    + 11 * (visit_type == "ICU").astype(float)
    + 6 * np.isin(department, ["ICU", "Cardiology"]).astype(float)
    + 6 * vitals_abnormal_flag
)
length_of_stay_hours = np.round(np.clip(los_base + rng.exponential(scale=6, size=N_VISITS), 0.5, 96), 2)

visits = pd.DataFrame({
    "visit_id": visit_id,
    "patient_id": visit_patient_id,
    "visit_date": visit_date,
    "department": department,
    "visit_type": visit_type,
    "length_of_stay_hours": length_of_stay_hours,
    "vitals_abnormal_flag": vitals_abnormal_flag,
    "risk_score": risk_score,
    "doctor_id": doctor_id,
})

# de-duplicate any incidental patient+date+department repeats from random generation
visits = visits.sort_values("visit_id").drop_duplicates(subset=["patient_id", "visit_date", "department"], keep="first")

# ─────────────────────────────────────────────────────────────────────────────
# 3. BILLING
# Injected signal for claim_status (Paid / Pending / Rejected):
#   Rejection risk rises with billed_amount and for two "stricter" insurance
#   providers (CareOne, MediCareX); Pending is more common for very recent
#   claims (short gap between visit and billing_date).
# billed_amount itself depends mildly on department + chronic_flag (realistic
# cost driver), then approved_amount / payment_days are derived CONSISTENTLY
# from claim_status (Paid -> full amount + payment turnaround, Rejected -> 0 +
# no payment_days, Pending -> partial amount + no payment_days) — same
# logically-safe rule used in the real cleaned dataset.
# ─────────────────────────────────────────────────────────────────────────────
visit_join = visits.merge(patients, on="patient_id")
N_BILL = len(visit_join)  # visits may be slightly fewer than N_VISITS after dedup
bill_id = np.arange(1, N_BILL + 1)

dept_base_cost = {"Cardiology": 26000, "Orthopedics": 22000, "ICU": 32000,
                   "General": 14000, "ER": 18000, "Neurology": 24000}
base_cost = visit_join["department"].map(dept_base_cost).values
billed_amount = np.round(
    base_cost * (1 + 0.15 * visit_join["chronic_flag"].values)
    * rng.uniform(0.75, 1.3, size=N_BILL),
    2,
)
billed_amount = np.clip(billed_amount, 500, 90000)

strict_provider = visit_join["insurance_provider"].isin(["CareOne", "MediCareX"]).astype(float)
billed_norm = billed_amount / billed_amount.max()
high_admin_dept = visit_join["department"].isin(["ICU", "ER"]).astype(float).values
patient_chronic = visit_join["chronic_flag"].values

# Two sequential threshold-based decisions (same style as risk_score's cut):
#   Step 1: Rejected vs not <- high billed_amount + strict insurance provider
#   Step 2 (within non-rejected): Pending vs Paid <- high-admin-overhead
#           department + non-chronic patient
# Threshold-based generation (rather than smooth probabilistic sampling)
# produces genuinely learnable decision boundaries for tree models, with a
# small amount of random label noise layered on top for realism.
score1 = 3.5 * billed_norm + 1.5 * strict_provider + rng.normal(0, 0.25, size=N_BILL)
reject_cutoff = np.quantile(score1, 0.85)  # ~15% Rejected, matching real-world rate
is_rejected = score1 > reject_cutoff

score2 = 2.5 * high_admin_dept + 1.3 * (1 - patient_chronic) + rng.normal(0, 0.25, size=N_BILL)
pending_cutoff = np.quantile(score2[~is_rejected], 0.70)  # ~26% of non-rejected -> Pending
is_pending = (~is_rejected) & (score2 > pending_cutoff)

claim_status = np.where(is_rejected, "Rejected", np.where(is_pending, "Pending", "Paid"))

# ~3% random label noise (keeps it realistic, avoids a suspiciously perfect model)
noise_mask = rng.random(N_BILL) < 0.03
noise_alt = rng.choice(["Paid", "Pending", "Rejected"], size=N_BILL)
claim_status = np.where(noise_mask, noise_alt, claim_status)

billing_gap_days = rng.integers(0, 30, size=N_BILL)  # days after estimated discharge

payment_days = np.where(
    claim_status == "Paid",
    np.round(rng.gamma(shape=3, scale=4, size=N_BILL) + 1, 0),
    np.nan,
)

approved_amount = np.select(
    [claim_status == "Paid", claim_status == "Rejected", claim_status == "Pending"],
    [billed_amount, 0.0, billed_amount * rng.uniform(0.5, 0.9, size=N_BILL)],
)

# estimated discharge -> billing_date always on/after it (logically consistent)
est_discharge = visit_join["visit_date"].values + (
    (visit_join["length_of_stay_hours"].values // 24).astype(int) + 1
).astype("timedelta64[D]")
billing_date = est_discharge + billing_gap_days.astype("timedelta64[D]")

billing = pd.DataFrame({
    "bill_id": bill_id,
    "visit_id": visits["visit_id"].values,
    "billed_amount": billed_amount,
    "approved_amount": np.round(approved_amount, 2),
    "claim_status": claim_status,
    "payment_days": payment_days,
    "billing_date": billing_date,
})

# ─────────────────────────────────────────────────────────────────────────────
# 4. Sanity checks (must all be zero — dates/logic are consistent by construction)
# ─────────────────────────────────────────────────────────────────────────────
check = visits.merge(patients, on="patient_id").merge(billing, on="visit_id")
assert (check["visit_date"] < check["registration_date"]).sum() == 0
assert (check["billing_date"] < check["visit_date"]).sum() == 0
assert ((check["claim_status"] != "Paid") & (check["payment_days"] > 0)).sum() == 0
print("Sanity checks passed: no illogical dates, no claim-status/payment_days conflicts.")

# ─────────────────────────────────────────────────────────────────────────────
# 5. Save
# ─────────────────────────────────────────────────────────────────────────────
for df, col in [(patients, "registration_date"), (visits, "visit_date"), (billing, "billing_date")]:
    df[col] = pd.to_datetime(df[col]).dt.strftime("%Y-%m-%d")

patients.to_csv("patients_synthetic.csv", index=False)
visits.to_csv("visits_synthetic.csv", index=False)
billing.to_csv("billing_synthetic.csv", index=False)

print("Saved patients_synthetic.csv, visits_synthetic.csv, billing_synthetic.csv")
print("\nrisk_score distribution:\n", pd.Series(risk_score).value_counts(normalize=True))
print("\nclaim_status distribution:\n", pd.Series(claim_status).value_counts(normalize=True))
```

```python
# ── Paths ─────────────────────────────────────────────────────────────────────
CSV_FILES = {
    "patients": Path("patients_synthetic.csv"),
    "visits":   Path("visits_synthetic.csv"),
    "billing":  Path("billing_synthetic.csv"),
}

# ── Column dtype mappings ─────────────────────────────────────────────────────
DTYPES = {
    "patients": {
        "patient_id":         "Int64",
        "age":                "Int64",
        "gender":             "str",
        "city":               "str",
        "insurance_provider": "str",
        "chronic_flag":       "Int64",
        "registration_date":  "str",
    },
    "visits": {
        "visit_id":             "Int64",
        "patient_id":           "Int64",
        "visit_date":           "str",
        "department":           "str",
        "visit_type":           "str",
        "length_of_stay_hours": "float",
        "risk_score":           "str",
        "doctor_id":            "Int64",
    },
    "billing": {
        "bill_id":         "Int64",
        "visit_id":        "Int64",
        "billed_amount":   "float",
        "approved_amount": "float",
        "claim_status":    "str",
        "payment_days":    "Int64",
        "billing_date":    "str",
    },
}

DATE_COLUMNS = {
    "patients": ["registration_date"],
    "visits":   ["visit_date"],
    "billing":  ["billing_date"],
}

# ── Functions ─────────────────────────────────────────────────────────────────
def load_csv(table: str) -> pd.DataFrame:
    path = CSV_FILES[table]
    if not path.exists():
        raise FileNotFoundError(f"CSV not found: {path}")

    df = pd.read_csv(path, dtype=DTYPES[table])

    for col in DATE_COLUMNS.get(table, []):
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors="coerce").dt.strftime("%Y-%m-%d")

    str_cols = df.select_dtypes(include=["object", "str"]).columns
    df[str_cols] = df[str_cols].apply(lambda s: s.str.strip())

    print(f"  Loaded {table}: {len(df):,} rows, {df.shape[1]} columns")
    return df


def create_schema(conn: sqlite3.Connection) -> None:
    conn.executescript("""
        CREATE TABLE IF NOT EXISTS patients (
            patient_id          INTEGER PRIMARY KEY,
            age                 INTEGER,
            gender              TEXT,
            city                TEXT,
            insurance_provider  TEXT,
            chronic_flag        INTEGER,
            registration_date   TEXT
        );

        CREATE TABLE IF NOT EXISTS visits (
            visit_id                INTEGER PRIMARY KEY,
            patient_id              INTEGER,
            visit_date              TEXT,
            department              TEXT,
            visit_type              TEXT,
            length_of_stay_hours    REAL,
            risk_score              TEXT,
            doctor_id               INTEGER,
            FOREIGN KEY (patient_id) REFERENCES patients(patient_id)
        );

        CREATE TABLE IF NOT EXISTS billing (
            bill_id          INTEGER PRIMARY KEY,
            visit_id         INTEGER,
            billed_amount    REAL,
            approved_amount  REAL,
            claim_status     TEXT,
            payment_days     INTEGER,
            billing_date     TEXT,
            FOREIGN KEY (visit_id) REFERENCES visits(visit_id)
        );

        CREATE INDEX IF NOT EXISTS idx_visits_patient   ON visits(patient_id);
        CREATE INDEX IF NOT EXISTS idx_billing_visit    ON billing(visit_id);
        CREATE INDEX IF NOT EXISTS idx_visits_date      ON visits(visit_date);
        CREATE INDEX IF NOT EXISTS idx_billing_status   ON billing(claim_status);
    """)
    conn.commit()
    print("  Schema applied.")


def insert_table(conn: sqlite3.Connection, table: str, df: pd.DataFrame) -> None:
    df.to_sql(table, conn, if_exists="replace", index=False)
    conn.commit()
    print(f"  Inserted {len(df):,} rows into '{table}'.")


def verify(conn: sqlite3.Connection) -> None:
    cur = conn.cursor()
    for table in ("patients", "visits", "billing"):
        cur.execute(f"SELECT COUNT(*) FROM {table}")
        print(f"  {table}: {cur.fetchone()[0]:,} rows")

    cur.execute("""
        SELECT
            p.patient_id, p.age, p.insurance_provider,
            v.visit_date, v.department, v.visit_type,
            b.billed_amount, b.claim_status
        FROM patients p
        JOIN visits   v ON v.patient_id = p.patient_id
        JOIN billing  b ON b.visit_id   = v.visit_id
        LIMIT 5;
    """)
    rows    = cur.fetchall()
    headers = [d[0] for d in cur.description]
    print("\n  Sample joined rows (patients ▶ visits ▶ billing):")
    print("  " + " | ".join(headers))
    print("  " + "-" * 80)
    for row in rows:
        print("  " + " | ".join(str(v) for v in row))


def main() -> None:
    print(f"\n{'='*60}")
    print("  Hospital Analytics — Database Loader")
    print(f"{'='*60}")

    if DB_PATH.exists():
        DB_PATH.unlink()
        print(f"  Removed existing database: {DB_PATH.name}")

    conn = sqlite3.connect(DB_PATH)
    conn.execute("PRAGMA foreign_keys = ON;")

    print("\n[1/3] Applying schema...")
    create_schema(conn)

    print("\n[2/3] Loading CSV files...")
    dataframes = {table: load_csv(table) for table in ("patients", "visits", "billing")}

    print("\n[3/3] Inserting data...")
    for table, df in dataframes.items():
        insert_table(conn, table, df)

    print("\n[Verification]")
    verify(conn)

    conn.close()
    print(f"\n  Database saved to: {DB_PATH}")
    print(f"{'='*60}\n")

try:
    conn.close()
    print("Previous connection closed.")
except NameError:
    print("No open connection found.")
main()
```

# Section-3: Operational Analysis

```python
def run(conn: sqlite3.Connection, label: str, sql: str) -> pd.DataFrame:
    """Execute a query, print the label, and return a DataFrame."""
    print(f"\n{'─'*60}")
    print(f"  {label}")
    print(f"{'─'*60}")
    df = pd.read_sql_query(sql, conn)
    print(df.to_string(index=False))
    return df


conn = sqlite3.connect(DB_PATH)
conn.execute("PRAGMA foreign_keys = ON;")
```

```python
# OA-1: Top 10 departments by total visit volume
run(conn, "OA-1 | Top 10 departments by visit volume", """
    SELECT
        department,
        COUNT(visit_id)  AS total_visits
    FROM visits
    GROUP BY department
    ORDER BY total_visits DESC
    LIMIT 10;
""")

# OA-2: Top 5 departments by average length of stay
run(conn, "OA-2 | Top 5 departments by avg length of stay (hours)", """
    SELECT
        department,
        ROUND(AVG(length_of_stay_hours), 2)  AS avg_stay_hours
    FROM visits
    WHERE length_of_stay_hours IS NOT NULL
      AND length_of_stay_hours >= 0
    GROUP BY department
    ORDER BY avg_stay_hours DESC
    LIMIT 5;
""")

# OA-3: Percentage of High Risk visits per department
# risk_score stores values like 'High', 'Medium', 'Low'
run(conn, "OA-3 | % High Risk visits per department", """
    SELECT
        department,
        COUNT(visit_id) AS total_visits,
        SUM(CASE WHEN risk_score = 'High' THEN 1 ELSE 0 END) AS high_risk_visits,
        ROUND(
            100.0 * SUM(CASE WHEN risk_score = 'High' THEN 1 ELSE 0 END)
            / COUNT(visit_id), 2
        ) AS high_risk_pct
    FROM visits
    GROUP BY department
    ORDER BY high_risk_pct DESC;
""")

# OA-4: Average visits per patient by city
run(conn, "OA-4 | Avg visits per patient by city", """
    SELECT
        p.city,
        COUNT(v.visit_id) AS total_visits,
        COUNT(DISTINCT p.patient_id) AS unique_patients,
        ROUND(
            1.0 * COUNT(v.visit_id) / COUNT(DISTINCT p.patient_id),
            2
        ) AS avg_visits_per_patient
    FROM patients p
    LEFT JOIN visits v ON p.patient_id = v.patient_id
    WHERE p.city IS NOT NULL
    GROUP BY p.city
    ORDER BY avg_visits_per_patient DESC;
""")

# OA-5: Doctors with most High Risk visits
# doctor_id is used (no doctor_name column in schema)
run(conn, "OA-5 | Doctors handling most High Risk visits", """
    SELECT
        doctor_id,
        COUNT(visit_id)  AS high_risk_visits
    FROM visits
    WHERE risk_score = 'High'
      AND doctor_id IS NOT NULL
    GROUP BY doctor_id
    ORDER BY high_risk_visits DESC
    LIMIT 10;
""")
```

# Section-4: Financial Analysis


*   Retrieving the top 10 insurance providers by total billed amount.
* Identifying the top 5 insurance providers with the highest claim rejection rate.
* Finding the average payment delay (payment_days) by insurance provider.
* Calculating the revenue realization ratio (approved_amount / billed_amount) by department.
* Identifying visits where billed_amount is high but approved_amount is zero or missing.



```python
# FA-1: Top 10 insurance providers by total billed amount
run(conn, "FA-1 | Top 10 insurance providers by total billed amount", """
    SELECT
        p.insurance_provider,
        ROUND(SUM(b.billed_amount), 2)  AS total_billed
    FROM billing b
    JOIN visits   v ON b.visit_id   = v.visit_id
    JOIN patients p ON v.patient_id = p.patient_id
    WHERE p.insurance_provider IS NOT NULL
    GROUP BY p.insurance_provider
    ORDER BY total_billed DESC
    LIMIT 10;
""")

# FA-2: Top 5 insurance providers by claim rejection rate
run(conn, "FA-2 | Top 5 insurance providers by claim rejection rate", """
    SELECT
        p.insurance_provider,
        COUNT(b.bill_id) AS total_claims,
        SUM(CASE WHEN b.claim_status = 'Rejected' THEN 1 ELSE 0 END) AS rejected_claims,
        ROUND(
            100.0 * SUM(CASE WHEN b.claim_status = 'Rejected' THEN 1 ELSE 0 END)
            / COUNT(b.bill_id), 2
        ) AS rejection_rate_pct
    FROM billing b
    JOIN visits   v ON b.visit_id   = v.visit_id
    JOIN patients p ON v.patient_id = p.patient_id
    WHERE p.insurance_provider IS NOT NULL
    GROUP BY p.insurance_provider
    ORDER BY rejection_rate_pct DESC
    LIMIT 5;
""")

# FA-3: Average payment delay by insurance provider
run(conn, "FA-3 | Avg payment delay (days) by insurance provider", """
    SELECT
        p.insurance_provider,
        ROUND(AVG(b.payment_days), 1)  AS avg_payment_days
    FROM billing b
    JOIN visits   v ON b.visit_id   = v.visit_id
    JOIN patients p ON v.patient_id = p.patient_id
    WHERE b.payment_days IS NOT NULL
      AND b.payment_days >= 0
      AND p.insurance_provider IS NOT NULL
    GROUP BY p.insurance_provider
    ORDER BY avg_payment_days DESC;
""")

# FA-4: Revenue realization ratio by department
run(conn, "FA-4 | Revenue realization ratio by department", """
    SELECT
        v.department,
        ROUND(SUM(b.billed_amount),   2) AS total_billed,
        ROUND(SUM(b.approved_amount), 2) AS total_approved,
        ROUND(
            SUM(b.approved_amount) / NULLIF(SUM(b.billed_amount), 0),
            4
        ) AS realization_ratio
    FROM billing b
    JOIN visits v ON b.visit_id = v.visit_id
    WHERE b.billed_amount IS NOT NULL
    GROUP BY v.department
    ORDER BY realization_ratio ASC;
""")

# FA-5: High billed amount but approved_amount is zero, NULL or missing
run(conn, "FA-5a | PAID claims — high billed, NULL or zero approved (revenue leakage)", """
    SELECT
        b.bill_id, b.visit_id, v.department, v.patient_id,
        p.insurance_provider, b.billed_amount,
        b.approved_amount,          -- show actual value (NULL or 0)
        b.claim_status
    FROM billing b
    JOIN visits   v ON b.visit_id   = v.visit_id
    JOIN patients p ON v.patient_id = p.patient_id
    WHERE b.claim_status = 'Paid'
      AND b.billed_amount > (SELECT AVG(billed_amount) FROM billing)
      AND (b.approved_amount IS NULL OR b.approved_amount = 0)
    ORDER BY b.billed_amount DESC;
""")
```
